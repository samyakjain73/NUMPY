```python
import numpy as np

np.random.seed(42)  # Makes the results reproducible

subjects = np.array(["Maths", "Physics", "Chemistry", "Computer Science", "English"])
marks = np.random.randint(0, 101, size=(50, 5))

print("Marks generated successfully!")
print("First 5 students' marks:\n", marks[:5])
```

```python
print("Shape:", marks.shape)
print("Dimensions:", marks.ndim)
print("Total values:", marks.size)
print("Data type:", marks.dtype)
print("Number of students:", marks.shape[0])
print("Number of subjects:", marks.shape[1])
```

```python
print("First 10 students:\n", marks[:10])
print("\nLast subject (English) marks:\n", marks[:, -1])
print("\nMarks in Maths and Physics for first 5 students:\n", marks[:5, 0:2])
```

```python
total_marks = marks.sum(axis=1)
average_marks = marks.mean(axis=1)

topper_index = np.argmax(total_marks)

print("Totals of first 10 students:", total_marks[:10])
print("Averages of first 10 students:", average_marks[:10])
print(f"Overall topper: Student {topper_index + 1}")
print("Topper total:", total_marks[topper_index], "/ 500")
print("Topper average:", round(average_marks[topper_index], 2))
```

```python
below_40 = marks < 40
above_90 = marks > 90

print("Number of subject marks below 40:", np.sum(below_40))
print("Number of subject marks above 90:", np.sum(above_90))
print("Marks above 90:", marks[above_90])

failed_students = np.any(marks < 40, axis=1)
print("Students with at least one subject below 40:", np.where(failed_students)[0] + 1)
print("Number of students with at least one failed subject:", np.sum(failed_students))
```

```python
grades = np.select(
    [average_marks >= 90, average_marks >= 75, average_marks >= 60,
     average_marks >= 40],
    ["A+", "A", "B", "C"],
    default="F"
)

for grade in ["A+", "A", "B", "C", "F"]:
    student_numbers = np.where(grades == grade)[0] + 1
    print(f"Grade {grade}: {len(student_numbers)} student(s) — {student_numbers}")
```

```python
subject_averages = marks.mean(axis=0)
subject_topper_indices = np.argmax(marks, axis=0)

for i, subject in enumerate(subjects):
    print(f"{subject}: average = {subject_averages[i]:.2f}, "
          f"highest mark = {marks[subject_topper_indices[i], i]} "
          f"(Student {subject_topper_indices[i] + 1})")
```

```python
passed = marks >= 40
pass_counts = passed.sum(axis=0)
pass_percentages = (pass_counts / marks.shape[0]) * 100

for i, subject in enumerate(subjects):
    print(f"{subject}: {pass_counts[i]}/50 passed "
          f"({pass_percentages[i]:.1f}%)")
```

```python
overall_average = marks.mean()
best_subject_index = np.argmax(subject_averages)
lowest_subject_index = np.argmin(subject_averages)

print("========== STUDENT MARKS REPORT ==========")
print("Students:", marks.shape[0])
print("Subjects:", marks.shape[1])
print(f"Overall average mark: {overall_average:.2f}")
print(f"Highest subject average: {subjects[best_subject_index]} "
      f"({subject_averages[best_subject_index]:.2f})")
print(f"Lowest subject average: {subjects[lowest_subject_index]} "
      f"({subject_averages[lowest_subject_index]:.2f})")
print("Students with at least one mark below 40:", int(failed_students.sum()))
print("Marks above 90:", int(above_90.sum()))
print("==========================================")
```
